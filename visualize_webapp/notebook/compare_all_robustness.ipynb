# Compare robustness (all optimizers / experiments)

Loads precomputed scoring checkpoints from [`compare_all_optimizers.ipynb`](compare_all_optimizers.ipynb) under `local/save/COMP/`. Each `period_robustness_rows` entry stores `d_tk` (same as `representation_similarity_factor` in [`scoring_one_optimizer.ipynb`](scoring_one_optimizer.ipynb)).

**X-axis**

- **`USE_TRAINING_SAMPLE_CHECKPOINT_X = False`**: subsample index `0 … N−1` on the x-axis (local index within each trajectory).
- **`USE_TRAINING_SAMPLE_CHECKPOINT_X = True`**: global **checkpoint index** along training. The layout matches the scorer subsample: the first **5** x-values are `t_start, t_start+1, …, t_start+4`; the remaining **N−5** are `numpy.linspace(t_start+5, t_end, N−5)` (inclusive endpoints), with `t_start` / `t_end` the first and last checkpoint index used for that period’s through-time curve (same construction as `get_period_checkpoint_indices` in `scoring_helpers`).

**Cheap bounds merge (`MERGE_TRAJ_CHECKPOINT_BOUNDS_FROM_RECOVERY`)**  
Old pickles may omit bounds. When this flag is `True` (default), each run reloads **recovery periods** and **`nts`** (needed for the same `acceleration` subsample as scoring) to rebuild the checkpoint list via `get_period_checkpoint_indices`, then writes `t_traj_start` / `t_traj_end` on each row—no saliency or representation recompute. Set `BTSP_START_STRATEGY` / `STRICT` to match [`compare_all_optimizers.ipynb`](compare_all_optimizers.ipynb) or the x-axis can misalign.

```python
import gzip
import pickle
import sys
import warnings
from collections import defaultdict
from pathlib import Path
from typing import Any, Iterator, cast

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns


def _find_project_root(start: Path | None = None) -> Path:
	p = (start or Path.cwd()).resolve()
	for cand in [p, *p.parents]:
		if (cand / "pyproject.toml").is_file():
			return cand
	return p


PROJECT_ROOT = _find_project_root()
if str(PROJECT_ROOT) not in sys.path:
	sys.path.insert(0, str(PROJECT_ROOT))

from visualize_webapp.app import (
	_NETWORK_SAMPLES_PER_DIGIT,
	_metrics,
	_nts,
	_sort_optimizer_ids,
)
from visualize_webapp.post_processing import _pp_dead, _pp_neuron_digit
from visualize_webapp.notebook.scoring_helpers import (
	filter_periods,
	get_iteration,
	get_n_assigned_digits,
	get_period_checkpoint_indices,
	get_recovery_periods,
	remove_inelligible_periods,
)

# --- behaviour (match compare_all_optimizers scoring cell) ---------------
BTSP_START_STRATEGY = "acceleration"
STRICT = False

# False: x = 0..N-1. True: x = global checkpoint index (see markdown).
USE_TRAINING_SAMPLE_CHECKPOINT_X = True

# Reload recovery periods + nts to attach t_traj_start / t_traj_end (no saliency / repr recompute).
MERGE_TRAJ_CHECKPOINT_BOUNDS_FROM_RECOVERY = True

SAVE_DIR = "local/save/COMP"


def rename_experiments(exp_name: str) -> str:
	if exp_name.startswith("digit_"):
		return " Control (no pretrain)"
	elif "mislabel" in exp_name:
		return "Recover"
	elif "relabel" in exp_name:
		return "Reinforce"
	else:
		return "Control"


def iter_saved_comp_results(save_dir: Path) -> Iterator[tuple[tuple[str, str], dict[str, Any]]]:
	"""Yield ``((experiment_id, optimizer_id), results_dict)`` for each COMP checkpoint on disk."""
	save_dir = Path(save_dir)
	if not save_dir.is_dir():
		return
	for d in sorted(save_dir.iterdir(), key=lambda x: x.name):
		if not d.is_dir():
			continue
		key_csv = d / "key.csv"
		res_gz = d / "results.pkl.gz"
		res_pkl = d / "results.pkl"
		if not key_csv.is_file() or (not res_gz.is_file() and not res_pkl.is_file()):
			continue
		key_df = pd.read_csv(key_csv)
		if len(key_df) != 1:
			continue
		eid, oid = str(key_df["eid"].iloc[0]), str(key_df["oid"].iloc[0])
		if res_gz.is_file():
			with gzip.open(res_gz, "rb") as f:
				data = pickle.load(f)
		else:
			with open(res_pkl, "rb") as f:
				data = pickle.load(f)
		yield (eid, oid), data


def training_sample_checkpoint_positions(t_start: float, t_end: float, n: int) -> np.ndarray:
	"""X positions for the N subsampled checkpoints along a BTSP period (training checkpoint index).

	First 5: ``t_start, t_start+1, …, t_start+4``. Remaining ``n - 5``: ``linspace(t_start+5, t_end, n-5)``
	(endpoints inclusive), matching the intended layout (5 dense + rest spread to ``t_end``).
	"""
	n = int(n)
	if n <= 0:
		return np.array([], dtype=float)
	ts, te = float(t_start), float(t_end)
	if n <= 5:
		return ts + np.arange(n, dtype=float)
	first = ts + np.arange(5, dtype=float)
	rest = np.linspace(ts + 5.0, te, n - 5, endpoint=True, dtype=float)
	return np.concatenate([first, rest])


def _period_bounds_from_row(r: dict) -> tuple[float | None, float | None]:
	"""Return (t_start, t_end) inclusive for the subsample layout; prefer merged trajectory bounds."""
	ts, te = r.get("t_traj_start"), r.get("t_traj_end")
	if ts is not None and te is not None:
		if isinstance(ts, float) and np.isnan(ts):
			ts = None
		if isinstance(te, float) and np.isnan(te):
			te = None
	if ts is not None and te is not None:
		try:
			return float(ts), float(te)
		except (TypeError, ValueError):
			pass
	st, en = r.get("start"), r.get("end")
	if st is None or en is None:
		return None, None
	if isinstance(st, float) and np.isnan(st):
		return None, None
	if isinstance(en, float) and np.isnan(en):
		return None, None
	try:
		return float(st), float(en)
	except (TypeError, ValueError):
		return None, None


_warned_missing_bounds = False
_warned_merge_skip = False


def enrich_period_rows_with_traj_checkpoint_bounds(out: dict[str, Any]) -> None:
	"""Set ``t_traj_start`` / ``t_traj_end`` (first/last checkpoint in the through-time subsample) on each row."""
	global _warned_merge_skip
	rows = out.get("period_robustness_rows")
	if not rows:
		return
	if out.get("error"):
		return
	eid = out.get("eid")
	mid = out.get("mid")
	oid = out.get("oid")
	rid = out.get("rid")
	if eid is None or mid is None or oid is None or rid is None:
		if not _warned_merge_skip:
			_warned_merge_skip = True
			warnings.warn(
				"COMP pickle lacks eid/mid/oid/rid; cannot merge trajectory checkpoint bounds.",
				UserWarning,
				stacklevel=1,
			)
		return

	metrics = _metrics(eid, mid, oid, rid=rid, w_cache=False)
	df_nd = cast(pd.DataFrame, _pp_neuron_digit(eid, mid, oid, rid=rid, w_cache=False))
	if df_nd is None or df_nd.empty:
		return
	df_nd = df_nd.copy()
	df_nd["n_active"] = df_nd["status"].apply(
		lambda s: 0
		if s == "inactive"
		else (_NETWORK_SAMPLES_PER_DIGIT if s == "assigned" else int(s.split("_")[-1]))
	)
	df_nd["iteration"] = df_nd["checkpoint_idx"].map(lambda c: get_iteration(metrics, c))

	df_dead = cast(pd.DataFrame, _pp_dead(eid, mid, oid, rid=rid, w_cache=False))
	if df_dead is not None and not df_dead.empty:
		df_dead = df_dead.copy()
		df_dead["iteration"] = df_dead["checkpoint_idx"].map(lambda c: get_iteration(metrics, c))

	nts = _nts(eid, mid, oid, rid=rid, w_cache=False)
	n_assigned_digits = get_n_assigned_digits(df_nd)
	recovery_periods = get_recovery_periods(
		metrics, df_nd, n_assigned_digits, df_dead, mode="assigned_same"
	)
	recovery_periods = remove_inelligible_periods(recovery_periods)
	if df_dead is not None and not df_dead.empty:
		recovery_periods = filter_periods(recovery_periods, df_dead, strict=STRICT)
	if recovery_periods.empty:
		return

	cp_df = get_period_checkpoint_indices(
		recovery_periods,
		stride=1,
		nts=nts,
		metrics=metrics,
		btsp_start_strategy=BTSP_START_STRATEGY,
	)
	if cp_df.empty:
		return

	bounds = (
		cp_df.groupby(["neuron_id", "period_id"], sort=False)["checkpoint_idx"]
		.agg(t_traj_start="min", t_traj_end="max")
	)
	b_df = bounds.reset_index()
	for r in rows:
		nid = r.get("neuron_id")
		pid = r.get("period_id")
		if nid is None or pid is None or (isinstance(pid, float) and np.isnan(pid)):
			continue
		sub = b_df[(b_df["neuron_id"] == nid) & (b_df["period_id"] == int(pid))]
		if sub.empty:
			continue
		r["t_traj_start"] = int(sub["t_traj_start"].iloc[0])
		r["t_traj_end"] = int(sub["t_traj_end"].iloc[0])


def period_robustness_rows_to_tt_df(rows: list[dict], *, use_training_axis: bool) -> tuple[pd.DataFrame, str]:
	"""Expand ``d_tk``; returns ``(df, x_col_name)`` for plotting."""
	global _warned_missing_bounds
	records: list[dict] = []
	for r in rows or []:
		d_tk = r.get("d_tk")
		if not isinstance(d_tk, (list, tuple)) or len(d_tk) == 0:
			continue
		n = len(d_tk)
		if use_training_axis:
			ts, te = _period_bounds_from_row(r)
			if ts is None or te is None:
				if not _warned_missing_bounds:
					_warned_missing_bounds = True
					warnings.warn(
						"period rows lack trajectory checkpoint bounds; using subsample indices on the x-axis. "
						"Set MERGE_TRAJ_CHECKPOINT_BOUNDS_FROM_RECOVERY=True or re-save COMP after merging bounds in compare_all_optimizers.",
						UserWarning,
						stacklevel=1,
					)
				xs = np.arange(n, dtype=float)
			else:
				xs = training_sample_checkpoint_positions(ts, te, n)
		else:
			xs = np.arange(n, dtype=float)
		for i, val in enumerate(d_tk):
			fv = float(val) if val is not None and not (isinstance(val, float) and np.isnan(val)) else np.nan
			row = {
				"neuron_id": r.get("neuron_id"),
				"period_id": int(r["period_id"]) if pd.notna(r.get("period_id")) else -1,
				"t_k_index": i,
				"representation_similarity_factor": fv,
			}
			if use_training_axis:
				row["checkpoint_idx"] = float(xs[i])
			records.append(row)
	x_col = "checkpoint_idx" if use_training_axis else "t_k_index"
	return pd.DataFrame.from_records(records), x_col


def plot_through_time_robustness(tt: pd.DataFrame, *, title: str, x_col: str) -> None:
	if tt.empty:
		print(f"(skip) no trajectory rows: {title}")
		return
	tt = tt.sort_values(["neuron_id", "period_id", x_col])
	tt["trace_id"] = tt["neuron_id"].astype(str) + " | p" + tt["period_id"].astype(str)
	fig, ax = plt.subplots(figsize=(14, 6))
	sns.lineplot(
		data=tt,
		x=x_col,
		y="representation_similarity_factor",
		hue="trace_id",
		estimator=None,
		units="trace_id",
		marker="o",
		markers=True,
		dashes=False,
		linewidth=0.9,
		alpha=0.7,
		legend=False,
		ax=ax,
	)
	ax.set_title(title)
	ax.set_xlabel("time (iteration)")
	ax.set_ylabel("$d(t_k)$")
	ax.grid(True, linestyle="--", alpha=0.4)
	plt.tight_layout()
	plt.show()


loaded = dict(iter_saved_comp_results(SAVE_DIR))
by_eid: dict[str, list[str]] = defaultdict(list)
for eid, oid in loaded:
	by_eid[eid].append(oid)
ordered_keys: list[tuple[str, str]] = []
for eid in sorted(by_eid.keys()):
	for oid in _sort_optimizer_ids(by_eid[eid]):
		ordered_keys.append((eid, oid))

print(f"Loaded {len(ordered_keys)} COMP result(s) from {SAVE_DIR}")

for (eid, oid) in ordered_keys:
	out = loaded[(eid, oid)]
	if MERGE_TRAJ_CHECKPOINT_BOUNDS_FROM_RECOVERY and USE_TRAINING_SAMPLE_CHECKPOINT_X:
		enrich_period_rows_with_traj_checkpoint_bounds(out)
	rows = out.get("period_robustness_rows") or []
	tt, x_col = period_robustness_rows_to_tt_df(rows, use_training_axis=USE_TRAINING_SAMPLE_CHECKPOINT_X)
	title = f"Through-time robustness: d(t_k) — {rename_experiments(eid)}  |  {oid}"
	plot_through_time_robustness(tt, title=title, x_col=x_col)
```

```python
# Boxplot: scalar ``robustness`` per neuron–period (one point per period in ``period_robustness_rows``).
# One x position per experiment; dodged boxes per optimizer (same COMP pickles as above).

from visualize_webapp.app import _opt_color, _opt_label

_rob_rows: list[dict] = []
for (eid, oid), out in loaded.items():
	for r in out.get("period_robustness_rows") or []:
		rv = r.get("robustness")
		if rv is None or (isinstance(rv, float) and np.isnan(rv)):
			continue
		_rob_rows.append(
			{
				"experiment_id": r.get("experiment_id", eid),
				"optimizer_id": r.get("optimizer_id", oid),
				"robustness": float(rv),
			}
		)

df_rob = pd.DataFrame.from_records(_rob_rows)
if df_rob.empty:
	print("No robustness rows to plot.")
else:
	df_rob["experiment"] = df_rob["experiment_id"].map(rename_experiments)
	exp_order = sorted(df_rob["experiment"].unique())
	opt_order = _sort_optimizer_ids(df_rob["optimizer_id"].unique().tolist())
	palette = [_opt_color(o) for o in opt_order]
	fig, ax = plt.subplots(figsize=(max(10, len(exp_order) * 1.4), 5.5))
	df_rob["robustness"] = 1 - df_rob["robustness"]
	sns.boxplot(
		data=df_rob,
		x="experiment",
		y="robustness",
		hue="optimizer_id",
		order=exp_order,
		hue_order=opt_order,
		palette=palette,
		dodge=True,
		ax=ax,
		linewidth=0.9,
		fliersize=2.5,
	)
	ax.set_xlabel("experiment")
	ax.set_ylabel("robustness")
	ax.set_title("Period feature robustness by experiment (one box per optimizer)")
	ax.tick_params(axis="x", rotation=22)
	handles, labels = ax.get_legend_handles_labels()
	ax.legend(
		handles,
		[_opt_label(lbl) for lbl in labels],
		title="optimizer",
		bbox_to_anchor=(1.02, 1),
		loc="upper left",
	)
	plt.tight_layout()
	plt.show()
```

```python
# Grouped bars: total number of BTSP periods per renamed experiment (``n_periods`` from each COMP run).
# One x cluster per experiment; one bar per optimizer (sums if several raw ``experiment_id`` map to the same label).

from visualize_webapp.app import _opt_color, _opt_label

_n_rows: list[dict] = []
for (eid, oid), out in loaded.items():
	raw_n = out.get("n_periods")
	if raw_n is None or (isinstance(raw_n, float) and np.isnan(raw_n)):
		n = len(out.get("period_robustness_rows") or [])
	else:
		n = int(raw_n)
	_n_rows.append(
		{
			"experiment": rename_experiments(eid),
			"optimizer_id": oid,
			"n_periods": n,
		}
	)

df_n = pd.DataFrame.from_records(_n_rows)
if df_n.empty:
	print("No period counts to plot.")
else:
	df_n = df_n.groupby(["experiment", "optimizer_id"], as_index=False)["n_periods"].sum()
	exp_order = sorted(df_n["experiment"].unique())
	opt_order = _sort_optimizer_ids(df_n["optimizer_id"].unique().tolist())
	palette = [_opt_color(o) for o in opt_order]
	fig, ax = plt.subplots(figsize=(max(9, len(exp_order) * 1.3), 5))
	sns.barplot(
		data=df_n,
		x="experiment",
		y="n_periods",
		hue="optimizer_id",
		order=exp_order,
		hue_order=opt_order,
		palette=palette,
		edgecolor="0.35",
		linewidth=0.6,
		ax=ax,
	)
	ax.set_xlabel("experiment")
	ax.set_ylabel("number of periods")
	ax.set_title("Period count by experiment and optimizer")
	ax.tick_params(axis="x", rotation=22)
	handles, labels = ax.get_legend_handles_labels()
	ax.legend(
		handles,
		[_opt_label(lbl) for lbl in labels],
		title="optimizer",
		bbox_to_anchor=(1.02, 1),
		loc="upper left",
	)
	plt.tight_layout()
	plt.show()
```

```python
# Boxplot: distribution of trajectory ``t_start`` (first checkpoint index in the through-time subsample) per period.
# One x tick per renamed experiment; one dodged box per optimizer. Ensures bounds are merged when that path is enabled.

from visualize_webapp.app import _opt_color, _opt_label

try:
	_merge_bounds = MERGE_TRAJ_CHECKPOINT_BOUNDS_FROM_RECOVERY
except NameError:
	_merge_bounds = False

if _merge_bounds:
	for out in loaded.values():
		enrich_period_rows_with_traj_checkpoint_bounds(out)

_tstart_rows: list[dict] = []
for (eid, oid), out in loaded.items():
	for r in out.get("period_robustness_rows") or []:
		ts = r.get("t_traj_start")
		if ts is None:
			continue
		if isinstance(ts, float) and np.isnan(ts):
			continue
		try:
			ts_f = float(ts)
		except (TypeError, ValueError):
			continue
		_tstart_rows.append(
			{
				"experiment": rename_experiments(eid),
				"optimizer_id": oid,
				"t_start": ts_f,
			}
		)

df_ts = pd.DataFrame.from_records(_tstart_rows)
if df_ts.empty:
	print("No t_start values: enable MERGE_TRAJ_CHECKPOINT_BOUNDS_FROM_RECOVERY and run the main cell, or ensure COMP rows carry t_traj_start.")
else:
	exp_order = sorted(df_ts["experiment"].unique())
	opt_order = _sort_optimizer_ids(df_ts["optimizer_id"].unique().tolist())
	palette = [_opt_color(o) for o in opt_order]
	fig, ax = plt.subplots(figsize=(max(10, len(exp_order) * 1.4), 5.5))
	sns.boxplot(
		data=df_ts,
		x="experiment",
		y="t_start",
		hue="optimizer_id",
		order=exp_order,
		hue_order=opt_order,
		palette=palette,
		dodge=True,
		ax=ax,
		linewidth=0.9,
		fliersize=2.5,
	)
	ax.set_xlabel("experiment")
	ax.set_ylabel("t_start (checkpoint index)")
	ax.set_title("Period trajectory t_start by experiment (one box per optimizer)")
	ax.tick_params(axis="x", rotation=22)
	handles, labels = ax.get_legend_handles_labels()
	ax.legend(
		handles,
		[_opt_label(lbl) for lbl in labels],
		title="optimizer",
		bbox_to_anchor=(1.02, 1),
		loc="upper left",
	)
	plt.tight_layout()
	plt.show()
```
